```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
def getImage(data, height, width):
    colorIndex = {0:[255,255,255], 1:[0,255,0],2:[255,0,0],3:[0,0,255],4:[255,182,193]}

    npimg = np.zeros((height, width, 3), dtype=np.uint8)
    for y in range(height):
        for x in range(width):
            npimg[y, x, :] = colorIndex.get(data[y][x], [0, 0, 0])
    return npimg
```

```python
directory="LP/gt"
rgb_directory="LP/rgb"
```

```python
if not os.path.exists(rgb_directory):
    os.makedirs(rgb_directory)
    
for filename in os.listdir(directory):
    file_path = os.path.join(directory, filename)
    if os.path.isfile(file_path) and file_path.endswith(".npy"):
        gt = np.load(file_path)
        img = getImage(gt, gt.shape[0], gt.shape[1])

        plt.imshow(img)
        plt.xticks([])
        plt.yticks([])
        plt.axis("off")
        plt.savefig(f"{rgb_directory}/{filename}.png", bbox_inches="tight", pad_inches=0, dpi=1200)

        print(f"Saved {file_path}.png")
```
