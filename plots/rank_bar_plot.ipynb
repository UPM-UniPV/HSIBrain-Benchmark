```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
```

```python
models = [
    "EfficientNet", "Hyb3D2D", "DBDA", "SSAN", "SpectralFormer", "HiT",  "RSSAN",
    "GhostNet", "ViT", "LiteDwNet"
]

scores_slimbrain = [65.08, 63.37, 61.83, 60.82, 60.37, 58.86, 57.61, 
                    56.69, 52.00, 36.44]
#errors_slimbrain = [21.98, 22.61, 22.50, 24.02, 22.77, 22.25, 23.33, 23.23, 15.07, 32.67]

models2 = [
    "Hyb3D2D", "DBDA", "RSSAN", "EfficientNet", "SSAN", "HiT",
    "SpectralFormer", "LiteDwNet", "GhostNet", "ViT"
]

categories_slimbrain = [
    "conv", "conv", "hyb", "hyb", "att",  "hyb", "hyb",
    "conv", "att", "conv"
]

scores_helicoid = [91.13, 90.51, 90.33, 89.72, 89.05, 88.29, 
                   87.57, 84.73, 84.22, 75.81]
#errors_helicoid = [16.06, 14.07, 12.66, 17.99, 15.95, 12.98, 18.82, 18.79, 22.14, 0]

color_map = {
    "conv": "#7AA3D9",
    "hyb": "#ABD19E",
    "att": "#FFE63A"
}

categories_helicoid = [
    "conv", "hyb", "hyb", "conv", "hyb", "hyb",
    "att", "conv", "conv", "att"
]
```

```python
mpl.rcParams['hatch.linewidth'] = 1

fig, axes = plt.subplots(1, 2, figsize=(12, 6), sharex=True)

# --- SLIMBRAIN ---
y = np.arange(len(models))
colors = [color_map[c] for c in categories_slimbrain]
#axes[0].barh(y, scores_slimbrain, xerr=errors_slimbrain, color=colors, height=0.6, capsize=4)
bars = axes[0].barh(y, scores_slimbrain, color=colors, height=0.6, capsize=4)
axes[0].set_yticks(y, labels=models)
axes[0].invert_yaxis()
axes[0].set_xlim(0, 100)
axes[0].set_xlabel("Average F1-score (%)")
axes[0].set_title("SLIMBRAIN Database")

axes[0].spines["top"].set_visible(False)
axes[0].spines["right"].set_visible(False)
axes[0].grid(axis="x", color="lightgray", linestyle="-", linewidth=0.7)
axes[0].set_axisbelow(True)

bars[0].set_hatch('//')
bars[0].set_edgecolor('#800080')

bars[8].set_hatch('//')
bars[8].set_edgecolor('#800080')

#for i, (val, err) in enumerate(zip(scores_slimbrain, errors_slimbrain)):#
for i, val in enumerate(scores_slimbrain):
    axes[0].text(val + 1, i, f"{val:.2f}", va="center", fontsize=8)

# --- HELICoiD ---
y2 = np.arange(len(models2))
colors2 = [color_map[c] for c in categories_helicoid]
#axes[1].barh(y2, scores_helicoid, xerr=errors_helicoid, color=colors2, height=0.6, capsize=4)
bars_hl = axes[1].barh(y2, scores_helicoid, color=colors2, height=0.6, capsize=4)

axes[1].set_yticks(y2, labels=models2)
axes[1].invert_yaxis()
axes[1].set_xlim(0, 100)
axes[1].set_xlabel("Average F1-score (%)")
axes[1].set_title("HELICoiD")

axes[1].spines["top"].set_visible(False)
axes[1].spines["right"].set_visible(False)
axes[1].grid(axis="x", color="lightgray", linestyle="-", linewidth=0.7)
axes[1].set_axisbelow(True)

bars_hl[3].set_hatch('//')
bars_hl[3].set_edgecolor('#800080')

bars_hl[9].set_hatch('//')
bars_hl[9].set_edgecolor('#800080')

#for i, (val, err) in enumerate(zip(scores_helicoid, errors_helicoid)):
#for i, (val, err) in scores_helicoid:
#    axes[1].text(val + err + 1, i, f"{val:.2f}", va="center", fontsize=8)

for i, val in enumerate(scores_helicoid):
    axes[1].text(val + 1, i, f"{val:.2f}", va="center", fontsize=8)

plt.tight_layout()
plt.savefig("outputs/rank_bar_plot.png", dpi=1200, bbox_inches='tight', transparent=True)
plt.show()
```
