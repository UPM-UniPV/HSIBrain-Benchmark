## SLIMBRAIN

```python
folder_cubi_sb = '../datasets/Madrid/hsi/'
files_cubi_sb = sorted([folder_cubi_sb+f for f in os.listdir(folder_cubi_sb) if f.endswith('.npy')])
```

```python
bande_25 = np.array([659.9532721069965, 670.6901278110637, 688.8987816319119, 702.5051985740749, 713.615281118321,
                     729.6524257344674, 740.6725793724072, 753.5695934015597, 768.1045986005809, 780.729784079351,
                     791.9978623012679, 807.3773715362006, 817.5107698177138, 830.181604265628 , 845.0284417296846,
                     855.4515077097817, 866.4974768585153, 880.2295227696292, 891.8857127447593, 899.8378701835213,
                     914.4796949756168, 922.0207800957263, 931.7406509120706, 940.9482263577094, 950.6398722361428])
```

```python
cube = np.load(files_cubi_sb[np.random.randint(0, len(files_cubi_sb))])
```

```python
print(cube.shape)
```

```python
cube_norm = (cube - cube.min()) / (cube.max() - cube.min())
```

```python
coords = {
    "x": np.arange(cube_norm.shape[0]),
    "y": np.arange(cube_norm.shape[1]),
    "wavelength": bande_25,
}

da = xr.DataArray(cube_norm, dims=("x", "y", "wavelength"), coords=coords).to_dataset(name="reflectance")
```

```python
cube_vis = hypercoast.image_cube(
    da,
    variable="reflectance",
    cmap="jet",
    clim=(0, 1),
    rgb_wavelengths=[950.6398722361428, 807.3773715362006, 670.6901278110637],
)
cube_vis.show()
```
